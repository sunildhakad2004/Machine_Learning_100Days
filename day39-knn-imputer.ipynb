```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv(r"Datasets\train.csv",usecols=['Age','Pclass','Fare','Survived'])
data.head(10)
```

```python
data.isnull().sum()
```

```python
x = data.drop('Survived',axis=1)
y = data['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
imputer = KNNImputer()
```

```python
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_imputed,y_train)
lr_pred = lr.predict(X_test_imputed)

accuracy = accuracy_score(y_test, lr_pred)
print(f"Accuracy of Logistic Regression with KNN Imputer: {accuracy:.4f}")
```

```python
imputer1 = SimpleImputer()

X_train_simple = imputer1.fit_transform(X_train_imputed)
X_test_simple = imputer1.transform(X_test_imputed)
```

```python
logistic_model = LogisticRegression()
logistic_model.fit(X_train_simple, y_train)
logistic_pred = logistic_model.predict(X_test_simple)

accuracy_simple = accuracy_score(y_test, logistic_pred)
print(f"Accuracy of Logistic Regression with Simple Imputer: {accuracy_simple:.4f}")
```
