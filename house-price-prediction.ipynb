```python
import pandas as pd
```

```python
data = {
    "Area": [800, 1000, 1200, 1500, 1800,
             2000, 2200, 2500, 2800, 3000,
             1600, 2400, 1100, 1900, 2700],
    "Bedrooms": [2, 2, 3, 3, 3, 4, 4, 4, 5, 5,
                 3, 4, 2, 3, 5],
    "Age": [15, 10, 8, 5, 12, 6, 10, 4, 7, 3,
            15, 8, 20, 9, 5],
    "Price": [35, 45, 58, 75, 82, 105, 110, 135,
              148, 165, 72, 125, 40, 90, 155]
}
```

```python
data = pd.DataFrame(data)
print(data)
```

```python
data.describe()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
x_train,x_test,y_train,y_test = train_test_split(data.drop('Price',axis=1),data['Price'],test_size=0.2,random_state=0)
```

```python
x_train
```

```python
y_test
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
predictions = model.predict(x_test)
print(predictions)
```

```python
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("R2 Score:", r2_score(y_test, predictions))
print("MAE:", mean_absolute_error(y_test, predictions))
```

```python
new_house = pd.DataFrame({
    "Area": [2000],
    "Bedrooms": [2],
    "Age": [20]
})

print("Predicted Price (₹ lakh):",
      model.predict(new_house)[0])
```
