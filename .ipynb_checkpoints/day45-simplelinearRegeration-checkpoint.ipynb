```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Datasets/placement1.csv")
data.head()
```

```python
# data index reset karne ke liye hum log reset_index() ka use karenge.
data = data.reset_index(drop=True)
data.head()
```

```python
data.info()
```

```python
x = data.iloc[:, 0:1]
y = data.iloc[:, -1]
```

```python
print(x)
```

```python
plt.scatter(data['cgpa'], data['package'])
plt.title('Scatter plot of CGPA vs Package')
plt.xlabel('CGPA')
plt.ylabel('Package')
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
x_test.head(5)
```

```python
y_test.head(5)
```

```python
print(model.predict(x_test.iloc[1].values.reshape(1,1)))
```

```python
plt.scatter(data['cgpa'], data['package'])
plt.plot(x_train, model.predict(x_train), color='red')
plt.title('Scatter plot of CGPA vs Package with Regression Line')
plt.xlabel('CGPA')
plt.ylabel('Package')
```

```python
b = model.intercept_
```

```python
m = model.coef_
```
